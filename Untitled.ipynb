```python
import os
import sys
import requests

API_KEY = os.getenv("OPENWEATHER_API_KEY")  # export OPENWEATHER_API_KEY=...
if not API_KEY:
    sys.exit("Не задан OPENWEATHER_API_KEY")

TIMEOUT = 10
GEO_URL = "http://api.openweathermap.org/geo/1.0/direct"
WEATHER_URL = "https://api.openweathermap.org/data/2.5/weather"


def get_coordinates(city: str) -> tuple[float, float] | None:
    try:
        r = requests.get(
            GEO_URL,
            params={"q": city, "limit": 1, "appid": API_KEY},
            timeout=TIMEOUT,
        )
    except requests.RequestException as e:
        print(f"Ошибка сети при геокодировании: {e}")
        return None

    if r.status_code != 200:
        print(f"Геокодер вернул {r.status_code}: {r.text[:200]}")
        return None

    try:
        geo = r.json()
    except ValueError:
        print("Геокодер вернул не-JSON.")
        return None

    if not isinstance(geo, list) or not geo:
        print(f"Город «{city}» не найден.")
        return None

    return geo[0]["lat"], geo[0]["lon"]


def get_weather(lat: float, lon: float) -> dict | None:
    try:
        r = requests.get(
            WEATHER_URL,
            params={
                "lat": lat,
                "lon": lon,
                "appid": API_KEY,
                "units": "metric",
                "lang": "ru",
            },
            timeout=TIMEOUT,
        )
    except requests.RequestException as e:
        print(f"Ошибка сети при запросе погоды: {e}")
        return None

    if r.status_code != 200:
        print(f"Погодный API вернул {r.status_code}: {r.text[:200]}")
        return None

    try:
        return r.json()
    except ValueError:
        print("Погодный API вернул не-JSON.")
        return None


def main() -> None:
    city = input("Введите город: ").strip()
    if not city:
        sys.exit("Город не введён.")

    coords = get_coordinates(city)
    if coords is None:
        return

    data = get_weather(*coords)
    if data is None:
        return

    print(f"Город: {data.get('name', city)}")
    print(f"Температура: {data['main']['temp']} °C")
    print(f"Погода: {data['weather'][0]['description']}")


if __name__ == "__main__":
    main()
    
```

```python
import requests  # подключаем библиотеку, которая умеет отправлять HTTP-запросы к сайтам

# Ключ доступа к OpenWeatherMap.
# Его выдаёт сайт после регистрации. В каждом запросе передаётся как appid,
# чтобы сервер понял, кто обращается, и дал ответ.
API_KEY = "ваш_ключ"

# Спрашиваем у пользователя город.
# input() показывает приглашение и читает строку.
# .strip() убирает лишние пробелы по краям — на случай, если ввели " Москва ".
city = input("Введите город: ").strip()


# ===================== ЗАПРОС 1: ГЕОКОДЕР =====================
# Геокодер — это сервис, который по названию города возвращает его координаты
# (широту и долготу). Без координат мы не сможем спросить погоду.

geo_response = requests.get(
    # Адрес геокодера: говорит серверу, что мы ищем координаты города.
    "http://api.openweathermap.org/geo/1.0/direct",

    # Параметры запроса. requests сам соберёт из них строку URL:
    #   ?q=Москва&limit=1&appid=ваш_ключ
    # q     — что искать (название города)
    # limit — вернуть максимум 1 результат
    # appid — наш ключ доступа
    params={"q": city, "limit": 1, "appid": API_KEY},

    # Если сервер не ответит за 10 секунд — прервать попытку.
    # Без этого программа могла бы «зависнуть» навсегда.
    timeout=10,
)


# Проверяем код ответа.
# 200 — всё хорошо, данные пришли.
# 401 — ключ неверный или ещё не активирован.
# 404 — неверный адрес.
# 429 — превышен бесплатный лимит запросов.
if geo_response.status_code != 200:
    # Печатаем код ошибки, чтобы понять, что именно сломалось.
    print(f"Ошибка геокодера: HTTP {geo_response.status_code}")

    # .text — тело ответа как строка. Первые 300 символов обычно содержат
    # причину ошибки (например, "Invalid API key"). Обрезаем, чтобы не заваливать консоль.
    print(geo_response.text[:300])

else:
    # .json() превращает JSON-текст в объекты Python:
    #   { } → словарь, [ ] → список, "текст" → строка, 12.5 → число.
    # Геокодер при успехе возвращает СПИСОК словарей:
    #   [ {"name": "Moscow", "lat": 55.75, "lon": 37.61} ]
    geo = geo_response.json()

    # Проверяем, что пришёл именно список и он не пустой.
    #   isinstance(geo, list) — true, если geo это список.
    #   not isinstance(...)    — true, если это НЕ список (например, словарь с ошибкой).
    #   not geo                — true, если список пустой (город не найден).
    # Без этой проверки следующий geo[0] упал бы с KeyError: 0.
    if not isinstance(geo, list) or not geo:
        print(f"Город «{city}» не найден. Ответ сервера: {geo}")

    else:
        # Берём первый (и единственный) элемент списка — найденный город.
        # Из него достаём значения по ключам "lat" и "lon".
        # Справа два числа, слева две переменные — Python раскладывает их сам.
        lat, lon = geo[0]["lat"], geo[0]["lon"]


        # ===================== ЗАПРОС 2: ПОГОДА =====================
        # Теперь, когда у нас есть координаты, спрашиваем погоду для этой точки.

        weather_response = requests.get(
            # Другой эндпоинт: отдаёт текущую погоду.
            "https://api.openweathermap.org/data/2.5/weather",

            params={
                "lat": lat,         # широта — где искать погоду
                "lon": lon,         # долгота
                "appid": API_KEY,   # ключ доступа
                "units": "metric",  # температура в °C (по умолчанию было бы в Кельвинах)
                "lang": "ru",       # описание погоды на русском языке
            },
            timeout=10,  # снова ограничиваем время ожидания
        )

        # Та же проверка статуса, что и раньше.
        if weather_response.status_code != 200:
            print(f"Ошибка погодного API: HTTP {weather_response.status_code}")
            print(weather_response.text[:300])

        else:
            # Превращаем JSON погоды в словарь Python.
            # Пример ответа:
            # {
            #   "name": "Moscow",
            #   "main": { "temp": 5.32, "feels_like": 2.1 },
            #   "weather": [ { "description": "облачно с прояснениями" } ]
            # }
            data = weather_response.json()

            # f-строки: всё, что в фигурных скобках, вычисляется и подставляется в текст.
            # Внимание: weather — это СПИСОК словарей,
            # поэтому сначала [0], потом ключ "description".

            print(f"Город: {data['name']}")                                  # "Moscow"
            print(f"Температура: {data['main']['temp']} °C")                 # 5.32
            print(f"Ощущается как: {data['main']['feels_like']} °C")         # 2.1
            print(f"Погода: {data['weather'][0]['description']}")            # описание
```
